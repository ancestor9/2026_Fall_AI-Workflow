## **1. 메모리 기반 MongoDB**

```python
# ----------------------------------------------------
# 1. 필요 라이브러리 설치 (mongomock 사용)
# ----------------------------------------------------
!pip install mongomock -q
```

```python
from datetime import datetime
from faker import Faker
import mongomock

# ----------------------------------------------------
# 2. Mock MongoDB 클라이언트 연결 (네트워크/보안 에러 100% 없음)
# ----------------------------------------------------
client = mongomock.MongoClient()

db = client["company_db"]
collection = db["employees"]

# ----------------------------------------------------
# 3. Faker 데이터 생성 및 저장
# ----------------------------------------------------
fake = Faker("ko_KR")
Faker.seed(42)

documents_to_insert = []

for i in range(1, 6):
    doc = {
        "emp_id": i,
        "name": fake.name(),
        "job": fake.job(),
        "contact": {"email": fake.email(), "phone": fake.phone_number()},
        "skills": fake.words(nb=3),
        "salary": fake.random_int(min=3000, max=8000, step=100),
        "created_at": datetime.now(),
    }
    documents_to_insert.append(doc)

# Document 저장
result = collection.insert_many(documents_to_insert)
print(f"✅ MongoDB Document 저장 성공! ({len(result.inserted_ids)}개 생성됨)\n")

# ----------------------------------------------------
# 4. 저장된 Document 조회
# ----------------------------------------------------
print("=== MongoDB 저장 데이터 조회 ===")
for doc in collection.find():
    print(doc)
```
